## 📌 1. Key Takeaways

- **Supervised Learning Paradigm**: Learning a mapping $f: X \to y$ from labeled pairs $(x_i, y_i)$.
- **Normal Equation**: $\mathbf{w}^* = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y} = \mathbf{X}^+ \mathbf{y}$ (via Pseudo-Inverse `np.linalg.pinv`).
- **Loss Function (MSE)**: $J(\mathbf{w}) = \frac{1}{2n} \| \mathbf{X} \mathbf{w} - \mathbf{y} \|^2_2$
- **4 Assumptions**: Linearity, Independence, Homoscedasticity, Normality.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Set root path for imports (navigating up 2 levels: notebooks/supervised -> ml-learning)
current_dir = Path.cwd()
ml_learning_dir = current_dir.parent.parent if current_dir.name in ["supervised", "unsupervised"] else current_dir.parent
src_dir = ml_learning_dir / "src"
if str(src_dir) not in sys.path:
    sys.path.append(str(src_dir))

from ml_learning.supervised.linear_regression import LinearRegression
from ml_learning.supervised.metrics import mean_squared_error, r2_score
from ai_journey_shared.plotting import setup_plot_style, save_figure
from ai_journey_shared.utils import set_seed

setup_plot_style()
set_seed(42)
print("✅ Supervised Learning environment initialized!")
```

## 🎲 2. Synthetic Data Generation

```python
n_samples = 100
X = np.linspace(0, 10, n_samples).reshape(-1, 1)
true_slope = 2.5
true_intercept = 4.0
noise = np.random.normal(0, 1.5, size=X.shape)
y = true_slope * X.ravel() + true_intercept + noise.ravel()

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.scatter(X, y, color="#1f77b4", alpha=0.7, label="Supervised Labeled Data")
ax.set_title("Supervised Learning: Linear Dataset")
ax.set_xlabel("Feature X")
ax.set_ylabel("Target y")
ax.legend()
plt.show()
```

## 💻 3. Model Training & Evaluation Walkthrough

```python
try:
    model = LinearRegression(fit_intercept=True)
    model.fit(X, y)
    y_pred = model.predict(X)

    print(f"🎯 Estimated Slope:     {model.coef_[0]:.4f}")
    print(f"🎯 Estimated Intercept: {model.intercept_:.4f}")
    print(f"📊 MSE Loss:            {mean_squared_error(y, y_pred):.4f}")
    print(f"📊 R^2 Score:           {r2_score(y, y_pred):.4f}")

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.scatter(X, y, color="#1f77b4", alpha=0.6, label="Data Points")
    ax.plot(X, y_pred, color="#d62728", linewidth=2.5, label="OLS Fit")
    ax.set_title("Supervised Linear Regression Fit")
    ax.legend()

    fig_path = ml_learning_dir / "figures" / "supervised_linear_fit.png"
    save_figure(fig, fig_path)
    print(f"🖼️ Saved fit plot to {fig_path}")

except NotImplementedError as e:
    print("⏳ Waiting for implementation in src/ml_learning/supervised/linear_regression.py")
    print(f"Detail: {e}")
```
