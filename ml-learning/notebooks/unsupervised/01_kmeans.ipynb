Welcome to the **Unsupervised Learning** module! In this notebook, we explore **K-Means Clustering**—one of the most popular unsupervised algorithms for grouping unlabeled data.

---

## 📌 1. Essential Knowledge & Key Takeaways (Kiến Thức Cần Nhớ)

### 🧠 Mathematical Formulation & Optimization Objective
Given an unlabeled dataset $X = \{x_1, x_2, \dots, x_n\}$, K-Means partitions the data into $K$ distinct clusters $C = \{C_1, C_2, \dots, C_K\}$.

The objective is to minimize the **Within-Cluster Sum of Squares (WCSS / Inertia)**:

$$J = \sum_{k=1}^K \sum_{x_i \in C_k} \| x_i - \mathbf{\mu}_k \|^2_2$$

where $\mathbf{\mu}_k = \frac{1}{|C_k|} \sum_{x_i \in C_k} x_i$ is the centroid (mean vector) of cluster $C_k$.

---

### 🔄 4 Steps of Lloyd's Algorithm
1. **Initialization**: Choose $K$ initial centroids $\mathbf{\mu}_1, \dots, \mathbf{\mu}_K$ randomly from $X$ using `np.random.choice(..., replace=False)`.
2. **Assignment (E-Step)**: Assign each point $x_i$ to the nearest centroid using vectorized Euclidean distance:
   $$c^{(i)} = \arg\min_{k} \| x_i - \mathbf{\mu}_k \|^2$$
3. **Update (M-Step)**: Recompute each centroid as the mean of all points assigned to that cluster:
   $$\mathbf{\mu}_k = \frac{1}{|C_k|} \sum_{i \in C_k} x_i$$
4. **Convergence Check**: Stop if centroids no longer change (`np.allclose(centroids, old_centroids, atol=tol)`) or if `max_iter` is reached.

---

### ⚡ 3 Critical Practical Considerations

1. **Feature Scaling is MANDATORY**: Because K-Means computes Euclidean distances $\sqrt{\sum (x_{ij} - \mu_{kj})^2}$, features with large scales (e.g. Salary in thousands) will completely dominate small-scale features (e.g. Age in years). Always use `StandardScaler`!
2. **Random Initialization Sensitivity**: K-Means can get stuck in suboptimal local minima depending on initial centroid choices. (Modern implementations use **K-Means++** initialization to spread initial centroids apart).
3. **Selecting Optimal $K$ (Elbow Method & Silhouette Score)**:
   - **Elbow Method**: Plot WCSS vs $K$, look for the point where WCSS reduction slows down drastically.
   - **Silhouette Score**: Measures cluster cohesion vs separation (range $[-1, 1]$, closer to $1$ is better).

## 🛠️ 2. Environment Initialization

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Add src directory to system path
current_dir = Path.cwd()
ml_learning_dir = current_dir.parent.parent if current_dir.name in ["supervised", "unsupervised"] else current_dir.parent
src_dir = ml_learning_dir / "src"
if str(src_dir) not in sys.path:
    sys.path.append(str(src_dir))

from ml_learning.unsupervised.kmeans import KMeans
from ai_journey_shared.plotting import setup_plot_style, save_figure
from ai_journey_shared.utils import set_seed

setup_plot_style()
set_seed(42)
print("✅ Unsupervised environment initialized with shared styles!")
```

## 🎲 3. Generate Synthetic 2D Clusters

```python
# Generate 3 distinct synthetic 2D clusters
cluster_1 = np.random.randn(100, 2) + np.array([0, 0])
cluster_2 = np.random.randn(100, 2) + np.array([6, 6])
cluster_3 = np.random.randn(100, 2) + np.array([0, 7])
X = np.vstack([cluster_1, cluster_2, cluster_3])

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(X[:, 0], X[:, 1], color="gray", alpha=0.6, label="Unlabeled Data Points")
ax.set_title("Unsupervised Clustering: Raw Feature Space (Unlabeled)")
ax.set_xlabel("Feature X1")
ax.set_ylabel("Feature X2")
ax.legend()
plt.show()
```

## 💻 4. Model Training & Cluster Visualization

```python
try:
    # Fit K-Means model with K = 3 clusters
    kmeans = KMeans(n_clusters=3, max_iter=300, tol=1e-4)
    kmeans.fit(X)
    labels = kmeans.predict(X)
    centroids = kmeans.centroids

    print(f"✅ Training finished! Final Centroids:\n{centroids}")

    # Plot cluster assignments & centroids
    fig, ax = plt.subplots(figsize=(8, 5))
    colors = ["#1f77b4", "#ff7f0e", "#2ca02c"]
    for k in range(3):
        cluster_points = X[labels == k]
        ax.scatter(cluster_points[:, 0], cluster_points[:, 1], color=colors[k], alpha=0.6, label=f"Cluster {k}")

    # Plot final centroids
    ax.scatter(centroids[:, 0], centroids[:, 1], color="red", marker="X", s=200, linewidths=2, edgecolor="black", label="Centroids")
    ax.set_title("K-Means Clustering Result & Centroids")
    ax.set_xlabel("Feature X1")
    ax.set_ylabel("Feature X2")
    ax.legend()

    fig_path = ml_learning_dir / "figures" / "kmeans_clustering_result.png"
    save_figure(fig, fig_path)
    print(f"🖼️ Saved clustering plot to {fig_path}")

except NotImplementedError as e:
    print("⏳ Waiting for implementation in src/ml_learning/unsupervised/kmeans.py")
    print(f"Detail: {e}")
```
