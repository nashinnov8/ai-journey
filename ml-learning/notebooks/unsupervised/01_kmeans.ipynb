## 📌 1. Key Takeaways

- **Unsupervised Learning Paradigm**: No targets $y$. The algorithm finds natural groupings in feature space $X$.
- **K-Means Objective**: Minimize Within-Cluster Sum of Squares (WCSS / Inertia):

$$J = \sum_{k=1}^K \sum_{x_i \in C_k} \| x_i - \mu_k \|^2$$

- **2-Step Iterative Optimization (Lloyd's Algorithm)**:
  1. **Expectation (Assignment)**: Assign each sample $x_i$ to nearest centroid $\mu_k$.
  2. **Maximization (Update)**: Recalculate centroid $\mu_k$ as mean of assigned samples.
